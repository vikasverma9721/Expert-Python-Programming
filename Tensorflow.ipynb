## Tensorflow Math Modules

```python
import tensorflow as tf
import numpy as np
```

```python
a = 10
b = 20
```

```python
a+b
```

```python
tf.add(a,b)
```

```python
# Addition of Two List
```

```python
l1 = [1,2,3,4]
l2 = [5,6,7,8]
tf.add(l1,l2)
```

```python
# Addition of Two tuple
t1 = (2,3,4,5)
t2 = (5,6,7,8)
tf.add(t1,t2)
```

```python
l1 = [1,2,3,4]
a = 10
tf.add(l1,a)
```

```python
# Given a tensor x and tensorr y, this operation x^y computes for corresponding elements in x and y
```

```python
x = tf.constant([[2,2],[3,3]])
y = tf.constant([[8,16],[2,3]])
tf.pow(x,y)
```

```python
# tf.math.abs

# real number
x = tf.constant([-2.25,3.25])
tf.abs(x)
```

```python
# complex number
x = tf.constant([[-2.25 + 4.75j], [-3.25 + 5.75j]])
tf.abs(x)
```

```python
# tf.math.subtract
x = [1,2,3,4,5]
y = 1
tf.subtract(x,y)
```

```python
x = [1,2,3,4,5]
y = 1
tf.subtract(y,x)
```

```python
# tf.math.multiply
x = tf.constant(([1,2,3,4]))
tf.math.multiply(x,x)
```

```python
tf.multiply(x,x)
```

```python
x = tf.ones([1,2]);
y = tf.ones([2,1]);
x * y
```

```python
# tf.math.maximum
x = tf.constant([0.,0.,0.,0.])
y = tf.constant([-2,0.,2.,5.])
tf.math.maximum(x,y)
```

```python
x = tf.constant([-5,0.,0.,0.])
y = tf.constant([-3.])
tf.math.maximum(x,y)
```

```python
# tf.math.minimum
x = tf.constant([0.,0.,0.,0.])
y = tf.constant([-5.,-2.,0.,3.])
tf.math.minimum(x,y)
```

```python
x = tf.constant([-5.,0.,0.,0.])
y = tf.constant(-3.)
tf.math.minimum(x,y)
```

```python
# tf.math.log
x = tf.constant([0,0.5,1,5])
tf.math.log(x)
```

```python
# tf.math.divide
x = tf.constant([10,20,30])
y = tf.constant([2,3,4])
tf.divide(x,y)
```

```python
# tf.math.pow
x = tf.constant([[2,2],[3,3]])
y = tf.constant([[8,16],[2,3]])
tf.pow(x,y)
```

```python
# tf.math.sqrt
x = tf.constant([[4.0],[16.0]])
tf.sqrt(x)
```

```python
y = tf.constant([[-4.0],[16.0]])
tf.sqrt(y)
```
