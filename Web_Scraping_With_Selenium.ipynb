```python
!pip install selenium
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

def run_selenium_demo():
    # Setting Chrome Options to Run Error-Free on the Cloud
    options = webdriver.ChromeOptions()
    options.add_argument('--headless')          # To hide the browser window
    options.add_argument('--no-sandbox')         # To bypass the security layer
    options.add_argument('--disable-dev-shm-usage') # for memory management

    print("Browser background me start ho raha hai...")
    driver = webdriver.Chrome(options=options)

    try:
        # Setting the screen size
        driver.set_window_size(1920, 1080)

        # Navigating Google.com
        print("Google.com open ho raha hai...")
        driver.get("https://www.google.com")
        time.sleep(2)

        # Locating the Search Box and Inserting Text
        print("Search box me text insert kiya ja raha hai...")
        search_box = driver.find_element(By.NAME, "q")

        # Entering text into a form
        search_box.send_keys("Python Selenium Tutorial")
        time.sleep(1)

        # Submit the form (press Enter)
        print("Search form submit ho raha hai...")
        search_box.send_keys(Keys.RETURN)
        time.sleep(3)
        print("Current URL after search:", driver.current_url)

        # Navigate to another website
        print("Ab Python official site par navigate kar rahe hain...")
        driver.get("https://www.python.org")
        time.sleep(3)
        print("Current URL:", driver.current_url)

        # Backward operation
        print("Moving Backward: Wapas search results par ja rahe hain...")
        driver.back()
        time.sleep(3)
        print("Current URL after backward:", driver.current_url)

        # Forward operation
        print("Moving Forward: Dobara Python site par ja rahe hain...")
        driver.forward()
        time.sleep(3)
        print("Current URL after forward:", driver.current_url)

        # Refreshing the page
        print("Page ko refresh kiya ja raha hai...")
        driver.refresh()
        time.sleep(2)

    except Exception as e:
        print(f"Oops! Koi error aayi: {e}")

    finally:
        # Completely quit/close the browser
        print("Saare basic operations poore ho gaye. Browser quit ho raha hai...")
        driver.quit()
        print("Success! Process complete.")
```
