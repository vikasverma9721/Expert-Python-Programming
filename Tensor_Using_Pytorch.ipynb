```python
z = torch.tensor([1.9,2.3,3.7,4.4])
```

```python
# round
torch.round(z)
```

- Rounding converts floating-point numbers (decimal numbers) within the tensor to their nearest integer. This is commonly referred to as rounding off.

```python
# ciel
torch.ceil(z)
```

- Its function is to round floating-point (decimal) numbers within the tensor to their next largest integer.

```python
# floor
torch.floor(z)
```

- This does the exact opposite of torch.ceil(). Its job is to round floating-point (decimal) numbers within the tensor to their nearest integer (integer) value.

```python
# clamp
torch.clamp(z, min=2,max=3)
```

- Clamp's job is to place a boundary (limit) on the numbers within the tensor. This is also known as "clip" or "capping" in simple language.

### ***Matrix Operations***

```python
d = torch.randint(size = (2,3), low = 0, high = 10)
e = torch.randint(size = (3,2), low = 0, high = 10)

print(d)
print(e)
```

```python
# matrix multiplacation
torch.matmul(d, e)
```

- The logic of matrix multiplication is similar to standard math—"Dot Product of Rows and Columns" (i.e. multiplying the horizontal row of the first matrix by the vertical column of the second matrix).

```python
vector1 = torch.tensor([1,2,3])
vector2 = torch.tensor([4,5,6])
```

```python
# dot product
torch.dot(vector1, vector2)
```

- It multiplies the opposite numbers of two equally sized vectors (1D tensors) together, and then adds the answers together. Ultimately, you get a single number.

```python
# transpose
torch.transpose(d,0,1)
```

- This function flips the matrix diagonally (corners). Data that was previously horizontal in rows is now vertical in columns.

```python
b = torch.randint(size=(3,3), low = 0, high = 10, dtype = torch.float32)
b
```

```python
# determinant
torch.det(b)
```

- It converts any square matrix (e.g. 2x2, 3x3) into a single number (scalar) by solving it according to mathematical rules.

```python
#inverse
torch.inverse(b)
```

### ***Comparison Oepration***

```python
a = torch.randint(size = (2,3), low = 0, high = 10)
b = torch.randint(size = (2,3), low = 0, high = 10)

print(a)
print(b)
```

```python
print("Greater Than (a > b):")
print(a > b)
print("--" * 30)

print("Less Than (a < b):")
print(a < b)
print("--" * 30)

print("Equal To (a == b):")
print(a == b)
print("--" * 30)

print("Not Equal To (a != b):")
print(a != b)
print("--" * 30)

print("Greater Than Equal To (a >= b):")
print(a >= b)
print("--" * 30)

print("Less Than Equal To (a <= b):")
print(a <= b)
```
